```python
''' 
Author: Khachaturyan Tigran
Assignment: #1
'''

# Define variables with comments specifying data types

gym_member = "Alex Alliton"  # String
preferred_weight_kg = 20.5  # Float
highest_reps = 25  # Integer
membership_active = True  # Boolean

# Define a dictionary with workout stats
# Dictionary with friends' names as keys and tuples of workout minutes as values
workout_stats = {
    "Alex": (30, 45, 20),  # Yoga, Running, Weightlifting
    "Jamie": (25, 50, 30),
    "Taylor": (40, 35, 25)
}

# Calculate total workout minutes and store them in a separate dictionary
total_minutes_dict = {}
for friend, minutes in workout_stats.items():
    total_minutes_dict[f"{friend}_Total"] = sum(minutes)

# Update workout_stats after iteration
workout_stats.update(total_minutes_dict)


# 2D list
gym_workout_list = []
for friend in workout_stats:
    if not friend.endswith("_Total"):
        minutes = list(workout_stats[friend])
        gym_workout_list.append(minutes)

#Extract and print yoga and running minutes for all friends
print("Yoga & Running minutes:", [row[:2] for row in gym_workout_list])

#Extract and print weightlifting minutes for the last two friends
print("Weightlifting minutes for last two friends:", [row[2] for row in gym_workout_list[-2:]])

# Checking if any friend's total workout minutes exceed 120
for friend, total_minutes in workout_stats.items():
    if isinstance(total_minutes, int) and total_minutes >= 120:
        print(f"Great job staying active, {friend.replace('_Total', '')}!")

# Allow user input to check a friend's workout details
friend_name = input("Enter a friend's name to check workout stats: ")
if friend_name in workout_stats:
    print(f"{friend_name}'s workout stats:")
    print(f"Yoga: {workout_stats[friend_name][0]} minutes")
    print(f"Running: {workout_stats[friend_name][1]} minutes")
    print(f"Weightlifting: {workout_stats[friend_name][2]} minutes")
    print(f"Total: {workout_stats.get(friend_name + '_Total', 'N/A')} minutes")
else:
    print(f"Friend {friend_name} not found in the records.")

# Determine the friend with the highest and lowest workout minutes
filtered_stats = {k: v for k, v in workout_stats.items() if k.endswith("_Total")}
most_active = max(filtered_stats, key=filtered_stats.get).replace('_Total', '')
least_active = min(filtered_stats, key=filtered_stats.get).replace('_Total', '')

print(f"Most active friend: {most_active} with {filtered_stats[f'{most_active}_Total']} minutes.")
print(f"Least active friend: {least_active} with {filtered_stats[f'{least_active}_Total']} minutes.")
```
